## Encapsulation

```python
class BankAccount:
    def __init__(self, balance, holderName):
        self.__balance=balance
        self.__holderName=holderName

    @property
    def balance(self):
        return self.__balance
    

    @property
    def holderName(self):
        return self.__holderName
```

```python
BA1 = BankAccount(5000000000, 'Max')

print(BA1.balance)
print(BA1.holderName)

```
