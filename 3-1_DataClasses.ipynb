# Dataclasses

```python
from dataclasses import dataclass

@dataclass
class Person:
    name: str
    age: int
    alive: bool = True
```

```python
p1 = Person('Max', 21)

print(p1)
```

```python
p2 = Person('Ella', 25)

print(p2)
```

```python
print(p1==p2)
```

```python
@dataclass
class Vehicle:
    manufacturer: str
    model: str
    variant: str
    modelYear: int

    def __post_init__(self):
        if self.modelYear<1888:
            raise ValueError("No 'vehicle' is older than 1888. ")
```

```python
c1 = Vehicle('Ford', 'Mustang', 'Mach 1', 2011)
# c2 = Vehicle('Benz', 'Patent Wagen', 0, 1880)
# >>ValueError: No 'vehicle' is older than 1888. 
```

```python
@dataclass(frozen=True)
class BankAccount:
    accNo:int
    accHolderName:str
    accHolderAge:int
    accBalance:float
```

```python
a1 = BankAccount(1, 'Max', 21, 5000000000)
```

```python
# a1.accHolderName = 'Ella'
# >>FrozenInstanceError: cannot assign to field 'accHolderName'
```

```python
from dataclasses import field


@dataclass(frozen=True)
class BankAccount2:
    accNo:int
    accHolderName:str
    accHolderAge:int
    accBalance:float
    accPassword: str = field(repr=False)
```

```python
a1 = BankAccount2(1, 'Max', 21, 5000000000, '5678yuigRTyug')

print(a1)
```
