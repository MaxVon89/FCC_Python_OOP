## Inheritance

```python
class Vehicle:

    def __init__(self, manufacturer, model, variant):
        self.manufacturer=manufacturer
        self.model=model
        self.variant=variant


    def vehStart(self):
        print("The vehicle has started. ")
        
    
    def vehStopped(self):
        print("The vehicle has stopped. ")
```

```python
class Car(Vehicle):

    def __init__(self, manufacturer, model, variant, noOfDoors):
        super().__init__(manufacturer, model, variant) #auto completed!
        self.noOfDoors=noOfDoors


    def rev(self):
        print("Grrrrrrr...")
```

```python
class Bike(Vehicle):

    def __init__(self, manufacturer, model, variant, noOfGears) :
        super().__init__(manufacturer, model, variant)
        self.noOfGears=noOfGears


    def wheely(self):
        print("Wheeeeeeeeeeee...")
```

```python
c1 = Car('Lexus', 'LX', '500', 4)
c1.vehStart()
c1.rev()
c1.vehStopped()
```

```python
b1 = Bike('Kawasaki', 'Ninja', 'R', 6)
b1.vehStart()
b1.wheely()
b1.vehStopped()
```
