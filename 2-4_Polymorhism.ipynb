```python
class Vehicle:

    def __init__(self, brand, model, var):
        self.brand=brand
        self.model=model
        self.var=var


    def start(self):
        print("Engine ON")


    def stop(self):
        print("Engine OFF")
```

```python
class Car(Vehicle):

    def __init__(self, brand, model, var, noOfDoors):
        super().__init__(brand, model, var)
        self.noOfDoors=noOfDoors


    def start(self):
        print("Car Engine ON")
        
    
    def stop(self):
        print("Car Engine OFF")
        
        

class Bike(Vehicle):

    def __init__(self, brand, model, var, noOfGears):
        super().__init__(brand, model, var)
        self.noOfGears=noOfGears


    def start(self):
        print("Bike Engine ON")
        
    
    def stop(self):
        print("Bike Engine OFF")
```

```python
vehicles = [
    Car('Lexus', 'LX', '500', 4),
    Bike('Kawasaki', 'Ninja', 'R', 6),
]
```

```python
def inspect(v):
    for vehicle in v:
        if isinstance(vehicle, Vehicle):
            print(f"Inspecting {vehicle.brand} {vehicle.model} {type(vehicle).__name__}")
            vehicle.start()
            vehicle.stop()

        else: raise Exception("Invalid object. ")


inspect(vehicles)
```

```python
class Aircraft(Vehicle):
    def __init__(self, brand, model, var, noOfEngines):
        super().__init__(brand, model, var)
        self.noOfEngines=noOfEngines

    
    def start(self):
        print("Aircraft Engine ON")
        
    
    def stop(self):
        print("Aircraft Engine OFF")
```

```python
vehicles.append(Aircraft('Boeing', '747', '8', 4))
```

```python
inspect(vehicles)
```
